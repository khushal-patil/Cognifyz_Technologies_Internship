Objective: Perform a geographical analysis of the restaurants in the dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import folium
import geopandas as gpd
import plotly.express as px
from folium.plugins import HeatMap, MarkerCluster
from IPython.display import IFrame
```

Step 1: Load the Dataset

```python
df = pd.read_csv(r"C:\\Users\\UNKNOWN_CODER\\Cognifyz_Technologies_Internship\\Task2\\Dataset.csv")
```

```python
df.head(5)
```

Step 2: Check for Missing Values

```python
print("Missing values per column:\n", df.isnull().sum())
```

```python
df.dropna(subset=["Latitude", "Longitude"], inplace=True)  # Remove missing coordinates
```

```python
display(df.info())
```

Step 3: Generate Interactive Map with Restaurant Locations

```python
map_center = [df['Latitude'].mean(), df['Longitude'].mean()]
map = folium.Map(location=map_center, zoom_start=10)
marker_cluster = MarkerCluster().add_to(map)

for _, row in df.iterrows():
    folium.Marker(
        [row['Latitude'], row['Longitude']],
        popup=f"{row['Restaurant Name']} ({row['Aggregate rating']})"
    ).add_to(marker_cluster)

map_file_path = 'restaurants_map.html'
map.save(map_file_path)
IFrame(map_file_path, width=800, height=600)
```

```python

# Grouping restaurants by city/locality
df['City'] = df['Locality'].fillna(df['City']) 

# Count restaurants per city
restaurant_counts = df['City'].value_counts().reset_index()
restaurant_counts.columns = ['City', 'Number of Restaurants']

# Calculate statistics: average rating, price range, and cuisine diversity
rating_stats = df.groupby('City')['Aggregate rating'].mean().reset_index()
price_stats = df.groupby('City')['Price range'].mean().reset_index()
cuisine_stats = df.groupby('City')['Cuisines'].nunique().reset_index()

# Merge all statistics
city_analysis = restaurant_counts.merge(rating_stats, on='City', how='left')

city_analysis.rename(columns={'Aggregate rating': 'Average Rating',
                              'Price range': 'Average Price Range',
                              'Cuisines': 'Unique Cuisines'}, inplace=True)

# Sort by number of restaurants
city_analysis = city_analysis.sort_values(by='Number of Restaurants', ascending=False)

# Display insights
print("Top Cities with the Most Restaurants:\n", city_analysis.head(10))
print("\nStatistical Overview:\n", city_analysis.describe())

# Plot the concentration of restaurants
plt.figure(figsize=(12, 6))
sns.barplot(data=city_analysis.head(15), x='City', y='Number of Restaurants', palette='viridis')
plt.xticks(rotation=45)
plt.title('Top 15 Cities with Most Restaurants')
plt.xlabel('City')
plt.ylabel('Number of Restaurants')
plt.show()
```
